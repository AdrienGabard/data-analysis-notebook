### Exercice 7️⃣ - Créer de nouvelles colonnes

Une classification importante pour les sources d'électricité est de savoir si elles sont bas-carbone ou haut-carbone. Vous pouvez créer de nouvelles colonnes avec la somme des colonnes pertinentes. Vous pouvez utiliser `pd.DataFrame.sum` (documentation [ici](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sum.html)). Faites attention à l'`axis` que vous devez choisir. À titre indicatif, vous pouvez considérer :

* Bas-carbone : `Nuclear`, `Wind`, `Solar`, `Hydraulic`, `Hydraulic pumping` (car il réduit l'électricité disponible par l'hydraulique) et `Bioenergies`
* Haut-carbone : `Fuel oil`, `Coal`, `Gas`

Une autre classification intéressante est la disponibilité de la source. Certaines sources peuvent être disponibles 24 heures sur 24, comme le nucléaire, mais d'autres sont intermittentes, comme l'éolien et le solaire. Vous pouvez utiliser la classification suivante :

* Haute disponibilité : `Fuel oil`, `Coal`, `Gas`, `Nuclear`, `Hydraulic`, `Hydraulic pumping` (car il réduit l'électricité disponible par l'hydraulique) et `Bioenergies`
* Intermittent : `Wind`, `Solar`

Pour l'instant, nous n'avons pas d'information pour classer les `Physical exchange`, vous pouvez donc les ignorer pour les deux classifications.

````python
simple_power_df["Low-carbon"] = simple_power_df[["Nuclear", "Wind", "Solar", "Hydraulic", "Hydraulic pumping", "Bioenergies"]].sum(axis="columns")
simple_power_df["High-carbon"] = simple_power_df[["Fuel oil", "Coal", "Gas"]].sum(axis="columns")
simple_power_df["High-availability"] = simple_power_df[["Fuel oil", "Coal", "Gas", "Nuclear", "Hydraulic", "Hydraulic pumping", "Bioenergies"]].sum(axis="columns")
simple_power_df["Intermittent"] = simple_power_df[["Wind", "Solar"]].sum(axis="columns")
````

### Exercice 8️⃣ - Rééchantillonner le dataframe

Il peut être pratique pour certains cas d'usage d'avoir les données sur une période plus longue : au lieu de toutes les 30 min, vous pouvez avoir des données journalières, mensuelles ou même annuelles. Vous pouvez le faire en utilisant `pd.DataFrame.resample` (documentation [ici](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.replace.html)).

Pour cette transformation, vous devez être prudent(e). Sommer des puissances dans ce cas peut être trompeur. Il vaut mieux sommer de l'énergie. En considérant que la puissance est constante sur chaque intervalle de 30 min, vous pouvez utiliser la formule suivante :

$$Energy = Power \cdot \Delta t$$

Rappelons que dans ce cas, $\Delta t = 0.5$, car notre intervalle entre les échantillons est de 30 min (0,5 heure). L'unité devient le MWh. Pour les données mensuelles et annuelles, vous pouvez diviser le résultat par 1e6 pour avoir des données en TWh.

````python
daily_energy_df = simple_power_df.resample("D", on="ds").sum()*0.5
monthly_energy_df = daily_energy_df.resample("ME").sum()
yearly_energy_df = monthly_energy_df.resample("YE").sum()

### Transform MWh into TWh

monthly_energy_df = monthly_energy_df/1e6
yearly_energy_df = yearly_energy_df/1e6
````

### 📈 Visualisation de données avec Plotly

[Plotly](https://plotly.com/python/getting-started/) est une bibliothèque de visualisation de données qui produit des graphiques interactifs avec très peu de lignes de code.

Il existe deux façons d'utiliser plotly : plotly express ou les objets graphiques (graph objects). Le premier vous permet de produire des graphiques avec très peu de lignes de code, mais avec une personnalisation limitée. Le second est un peu plus verbeux, mais permet plus de détails.

Avec plotly express, vous choisissez essentiellement le type de graphique et donnez comme arguments le dataframe et les colonnes pour les axes x et y.

Créons par exemple un graphique en barres (documentation [ici](https://plotly.com/python/bar-charts/)) pour visualiser l'évolution de la consommation d'énergie annuelle d'année en année.

#### Exercice 3️⃣ - Améliorer le graphique en barres

Vous pourriez vouloir améliorer un peu ce graphique : peut-être ajouter un titre, définir une hauteur et une largeur différentes, modifier les étiquettes des axes, notamment l'axe y pour indiquer les unités, et également modifier le modèle de survol (vérifiez cet exemple [ici](https://plotly.com/python/hover-text-and-formatting/#modifying-the-hovertemplate-of-a-plotly-express-figure)).

#### Exercice 7️⃣ - Préparer les jeux d'entraînement et de test

Pour évaluer la qualité d'ajustement d'un modèle, vous devez le tester avec des données que le modèle n'a pas utilisées pour s'entraîner. C'est pourquoi vous devez diviser les données en jeu d'entraînement et jeu de test.

Comme indiqué précédemment, vous entraînerez votre modèle avec les données historiques de 2012 à 2021 et utiliserez les données de 2022 pour évaluer les performances de votre modèle.

Par conséquent, votre jeu de test sera constitué des 12 derniers points de données (toute l'année 2022). Vous pouvez utiliser `pd.DataFrame.tail` (documentation [ici](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.tail.html)). Cela signifie que pour le jeu d'entraînement, vous devez "laisser de côté" les 12 derniers points. #### Exercice 8️⃣ - Entraîner un modèle AutoARIMA

Dans un projet de Data Science, de nombreux modèles sont normalement testés et comparés les uns aux autres. Pour les besoins de ce notebook, vous n'utiliserez qu'un seul modèle, appelé AutoARIMA. Il combine l'auto-régression, la moyenne mobile et l'intégration pour réaliser une prévision. Vous n'avez rien compris à tout ça ? Pour l'instant, pas de panique. L'objectif de ce notebook est de se familiariser avec les bibliothèques Python et de voir ce que nous pouvons faire avec elles. Nous ne voulons pas devenir des experts en AutoARIMA (pas encore 😀). Entraînons un modèle AutoARIMA avec statsforecast.

Tout d'abord, vous devez créer une instance `StatsForecast`, que vous avez importée au début de ce notebook.

Dans cette instance, vous devez lister les modèles que vous souhaitez utiliser. Dans notre cas, AutoARIMA. Le modèle AutoARIMA accepte la saisonnalité. Dans notre cas, il y a une saisonnalité annuelle pour la consommation d'électricité, elle doit donc être définie à 12 (1 an = 12 mois). Il demande également la fréquence des données, qui est mensuelle (`ME` ou "fin de mois")

````python
sf = StatsForecast(
    models=[AutoARIMA(season_length = 12)],
    freq='ME',
)
sf.fit(train_set)
````

#### Exercice 9️⃣ - Réaliser une prévision

Pour réaliser une prévision, vous pouvez utiliser la méthode `predict` de l'instance StatsForecast, qui prend deux arguments :
* `h` : l'horizon de prévision. Vous souhaitez prévoir pour 12 mois.
* `level` : calcule un intervalle de prédiction. Par exemple, level=[95] signifie que le modèle s'attend à ce que la valeur réelle se trouve dans cet intervalle 95% du temps. Définissons-le à ce niveau.

````python
forecast_df = sf.predict(h=12, level=[95])
````

#### Exercice 🔟 - Calculer le MAPE des prévisions

Le MAPE est l'Erreur Absolue Moyenne en Pourcentage (Mean Absolute Percentage Error). Si nous avons des valeurs réelles $Y$ et des valeurs prévues $\hat{Y}$, le MAPE est calculé comme suit :

$$ MAPE = \frac{1}{n}\sum_{i=1}^n \left\lvert{\frac{Y - \hat{Y}}{Y}}  \right\rvert $$

où $n$ est la longueur des vecteurs $Y$ et $\hat{Y}$. Calculons-le étape par étape :

````python
Y = test_set["y"].values
Y_hat = forecast_df["AutoARIMA"].values
pc_error = (Y - Y_hat)/(Y)
abs_pc_error = np.abs(pc_error)
MAPE = np.mean(abs_pc_error)
print(MAPE)
````

Notre MAPE est de 4,29%. C'est peut-être un peu trop pour RTE, mais encore une fois, c'est l'erreur d'un modèle qui n'a pas été optimisé. De plus, comme nous l'avons vu au début, RTE s'intéresse à une prévision plus régulière, avec une période plus courte, et non mensuelle.

Un point important à noter pour le MAPE est que, comme il s'agit d'une erreur moyenne, il ne capture pas la variation des erreurs. Vous pouvez facilement remarquer que l'erreur pour juillet est très faible (0,34%), mais que l'erreur pour octobre est très élevée (14,6%).
